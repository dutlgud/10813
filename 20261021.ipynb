```python
hour=int(input('시간을 입력하세요:'))
minute=int(input('분을 입력하세여:'))
france_hour=hour-7
if france_hour<0:
  france_hour+=24
print('우리나라시간은 %d시%d분입니다.'%(hour, minute))
print('프랑스시간은 %d시%d분입니다.'%(france_hour, minute))
```
