```python
from gtts import gTTS
text="안녕하세요.gTTS를 이용해 텍스트를 음성으로 뱐환하고 있습니다."
tts=gTTS(text=text, lang="ko")
tts.save("output.mp3")
print("음성 파일 저장 완료!")
```

```python
from gtts import gTTS
from IPython.display import Audio
while True:
  text = input('텍스트틀 입력하세요:')
  if text =='종료':
    print('TTS 종료')
    break
  tts=gTTS(text, lang='en')
  tts.save('output.mp3')
  display(Audio('output.mp3',autoplay=True))
```
