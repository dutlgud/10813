```python
import random
mem=['김민준','이서준','박도윤','최시우','정예준','정예준','강주원','조유준']
winners_cnt=int(input('뽑을 당첨자 수를 입력하세요:'))
winers=random.sample(mem, winners_cnt)
winers.sort()
print(winers)
```
