```python
a=6
b=1.5
c=1+2j
print(a, type(a))
print(b, type(b))
print(c, type(c))
print(a+1, type(a+1)) #정수 + 정수=정수
print(a/b, type(a/b)) #정수 /실수=정수
print(a/2, type(a/2)) #정수 / 정수=실수
print(a//2, type(a//2)) #정수// 정수=정수
```

```python
# 한문자만 인덱싱 : 변수명[인덱스번호]
a= 'Hello Python'
print(a[0])
print(a[6])
print(a[-1])
```

```python
#연속된 부분의 문자자열로 슬라이싱 : 변수명[시작인덱스:종료인덱스]
a='Hello Python'
print(a[0:5])
print(a[0:6])
print(a[:5])
print(a[6:-4])
print(a[0:9:2])
```
